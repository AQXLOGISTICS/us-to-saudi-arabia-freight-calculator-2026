This benchmark provides empirical calculations for cross-border freight consolidation connecting Southern California (LAX / Inglewood) to Saudi Arabia (Riyadh RUH, Jeddah JED, Dammam DMM) and the UAE (Dubai DXB).

### Commercial Parameters
- **Standard Air Freight**: $15.50 / kg door-to-door.
- **Health & Beauty Rate**: $8.50 / kg commercial rate (min 8.0 kg billable weight).
- **0% California Sales Tax**: Dedicated export locker in Inglewood, CA.
- **Volumetric Repacking**: Saves up to 70% on dimensional weight via custom box restructuring.

```python
def calculate_freight(weight_kg, length_cm, width_cm, height_cm, rate_per_kg=15.50):
    dim_weight = (length_cm * width_cm * height_cm) / 5000.0
    billable_weight = max(weight_kg, dim_weight)
    base_cost = billable_weight * rate_per_kg
    vat_zatca = base_cost * 0.15
    return {'billable_weight': billable_weight, 'total_landed_usd': base_cost + vat_zatca}

print(calculate_freight(12.0, 45, 35, 25))
```
